```python
import math

def calculate_eoq(demand, order_cost, holding_cost):
    if demand <= 0 or order_cost <= 0 or holding_cost <= 0:
        raise ValueError("All inputs must be positive numbers.")
    return math.sqrt((2 * demand * order_cost) / holding_cost)

def eoq_main():
    try:
        demand = float(input("Enter annual demand (units/year): "))
        order_cost = float(input("Enter ordering cost ($/order): "))
        holding_cost = float(input("Enter annual holding cost per unit ($/year/unit): "))

        eoq = calculate_eoq(demand, order_cost, holding_cost)
        print(f"Economic Order Quantity (EOQ): {eoq:.2f} units")

    except ValueError as e:
        print(f"Error: {e}")

eoq_main()
```

```python
import random

play= input("Do you want to play a game (y/n?)")

while play.lower() == "y":
  player_choice= input("Choose b/w rock, paper, scissors: ").lower()
  computer_choice= random.choice(['rock', 'paper', 'scissors'])

  print(f"Player: {player_choice}")
  print(f"Computer: {computer_choice}")

  if player_choice == computer_choice:
    print("Draw")
  elif (
    (player_choice == "rock" and computer_choice == "scissors") or
    (player_choice == "paper" and computer_choice == "rock") or
    (player_choice == "scissors" and computer_choice == "paper")
  ):
    print("Player wins!")
  elif player_choice not in ['rock', 'paper', 'scissors']:
    print("Invalid choice. Please choose rock, paper, or scissors.")
  else:
    print ("Computer wins!")

  play= input("Do you want to play a game (y/n?)")
```

```python
def calcFee(hours):
  base_fee = 5.0
  hourly_rate = 2.50
  min_fee = 10.0
  max_fee = 20.0
  fee = base_fee + (hourly_rate * hours)
  if fee < min_fee:
    fee = min_fee
  elif fee > max_fee:
    fee = max_fee
  return fee

def main():
  try:
    hours_input = int(input("Enter number of parking hours:"))
    if hours_input < 0:
      print("Parking hours cannot be negative. Please enter a positive number.")
      return
    parking_fee = calcFee(hours_input)
    print(f"Parking Fee for {hours_input} hours: ${parking_fee:.2f}")
  except ValueError:
    print("Invalid input. Please enter a valid number for hours.")

main()
```

```python
def loan_payment(principal, annual_rate, months):
  assert(isinstance(principal, (int, float)) and principal >= 0)
  assert(isinstance(annual_rate, (int, float)) and annual_rate >= 0)
  assert(isinstance(months, (int, float)) and months > 0)

  try:
    monthly_rate = annual_rate / 12
    if monthly_rate == 0:
      monthly_payment = principal / months
    else:
      monthly_payment = (principal * monthly_rate * (1 + monthly_rate)**months) / (((1 + monthly_rate)**months) - 1)
    return monthly_payment
  except ZeroDivisionError:
    raise ValueError("Cannot calculate payment with zero months or invalid rate.")
  except Exception as e:
    raise ValueError(f"An unexpected error occurred: {e}")
```

```python
def getValues(n):
    assert isinstance(n, int) and n > 0, "n must be a positive integer"
    values = []
    for i in range(n):
        values.append(float(input(f"Enter value {i+1}: ")))
    return values[2:]


def calcAverage(alist):
    assert isinstance(alist, list), "Input parameter 'alist' must be a list."
    if not alist:
        raise ValueError("Input list for calcAverage cannot be empty.")
    return sum(alist) / len(alist)



try:
    n_input = input("Enter number of items to be added to the list: ")
    n = int(n_input)

    if n < 3:
        raise ValueError("Number of items must be at least 3 to generate values for averaging (after removing the first two).")

    alist = getValues(n)

    avg = calcAverage(alist)
    print(f"Average: {avg} lbs.")

except ValueError as ve:
    print(f"Error: {ve}")
except AssertionError as ae:
    print(f"Assertion Error: {ae}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def distance(self, other):
    dx = (self.x - other.x) ** 2
    dy = (self.y - other.y) ** 2
    return (dx + dy) ** 0.5

class Rectangle(Point):
  def __init__(self):
    self.p1 = Point(0,0)
    self.p2 = Point(5,0)
    self.p3 = Point(0,5)
    self.p4 = Point(5,5)
    self.width = self.p1.distance(self.p2)
    self.height = self.p1.distance(self.p3)

  def area(self):
    return self.width * self.height
```

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.wage = wage
        self.hours = hours

    def calculate_pay(self):
        return self.wage * self.hours

    def __str__(self):
        return f"{self.name}: wage={self.wage}, hours={self.hours}"

    def get_wage(self):
        return self.wage

    def set_wage(self, new_wage):
        self.wage = new_wage


class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):
        return (1 + self.bonus) * self.get_wage() * self.hours


employee = Employee("Bob", 50, 10)
manager = Manager("Alica", 100, 15, 0.2)

print(employee)
print(manager)
print()
print(f"Employee {employee.name} weekly payment is {employee.calculate_pay()}$")
print(f"Manager {manager.name} weekly payment is {manager.calculate_pay()}$ that includes a {manager.bonus * 100}% bonus.")
```
